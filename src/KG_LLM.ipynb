## Create Node embedding and search

```python

from langchain_neo4j import Neo4jVector
from llm import embedding
from common import NeoDB

vector_index = Neo4jVector.from_existing_graph(
    embedding= embedding,
    username=NeoDB().username,
    url = NeoDB().url,
    password=NeoDB().pwd,
    database = "fta2023",
    index_name='vi_port',
    node_label= 'Airport',
    text_node_properties=['name','desc','<id>'],
    embedding_node_property='embed_desc')

response=vector_index.similarity_search(
    'Which airport is located in Kerala?'
)
print(response[0].page_content)

```
