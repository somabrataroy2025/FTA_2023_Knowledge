## Higher order function

```python
def apply(func,x):
    return func(x)

print(apply(lambda p: p * 5,10))
```

## lambda

```python
add = lambda p,q,r: ((p+q)/r) * 7
print(add(2,3,4))
```

## map, reduce, filter

```python
import functools
data = [4,6,8,12]
dub = [2]
z = 2
res = list(map(lambda x :  x/2,data))
res1 = functools.reduce(lambda x,y: x*y,data)
res2 = functools.reduce(lambda x,y: x*y*z,data,z)

print(res1)
print(res2)

def mult(x): 
    return x == 8
res = filter(mult,data)
print(list(res))
```

# decorator

```python
def add_exclamation(func):
    print('Call1')
    def wrapper(name):
        print(f'{func.__name__}')
        return func(name)+ ' !'
    print('call3')
    return wrapper

@add_exclamation
def greet(inp:str):
    return inp

print(greet('Somabrata'))
    
```

## generator

```python
def numGenerator():
    yield 'a'
    yield 'b'
    yield 'c'

for num in numGenerator():
    print(num,end=' end ')
```

## *args **kwargs

```python
def display_info(name, *args, **kwargs):
    print("Name:", name)
    print("Additional Info:", args)
    print("Keyword Info:", kwargs)


display_info("John", 28,21,44, Role="Engineer", location="NY", hobby="Cycling")
```

## Comprehension (List, Dict, Set, Generator)

```python
lst = [1,2,3,4,5,6,7]
squares = [x * 2 for x in lst]
print(squares)

dct = {x : x%2==0  for x in lst}
print(dct)
```

```python
class abc:
    greet='Hello'

    @classmethod
    def Say(cls):
        return cls.greet
    
print(abc.Say())
```

```python
class Test:
    def __init__(self,name):
        self.name = name
    def __str__(self):
        return f"{self.name}"
    def Call(self):
        print(self.name)
    def __repr__(self):
        return f"obj is : {self.name}"
    def __enter__(self):
        print('Context Entered')
    def __exit__(self):
        print('exit')
```

```python
class check:
    __slots__ = ['age','name']
    def __init__(self):
        self.name = 'som'
        self.age = 40
```

```python
class Temperature:
    def __init__(self, celsius):
        self._celsius = celsius

    @celsius
    def celsius(self):
        return self._celsius
```
